```python
import pandas as pd

dataset = pd.read_csv("../../Output Datasets/final_features_otu_summary.csv")
dataset = dataset.transpose()
dataset.columns = dataset.iloc[0]
dataset.columns.name = None
dataset = dataset[1:]
dataset = dataset.reset_index()
dataset.rename(columns={dataset.columns[0]: "OTU"}, inplace=True)
dataset.head()
```

```python
taxa = pd.read_excel("../../Output Datasets/Final Selected OTUs.xlsx")
taxa.head()
```

```python
merged_df = pd.concat([dataset, taxa], axis=1)
merged_df.head()
```

```python
merged_df.to_csv("../../Output Datasets/final_features_otu_summary_merged.csv", index=False)
```
